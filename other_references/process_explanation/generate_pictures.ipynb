```python
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch, Ellipse
import numpy as np
import os

OUTPUT_DIR = os.path.join(os.getcwd(), 'output')
os.makedirs(OUTPUT_DIR, exist_ok=True)

# — Shared color palette —
COLORS = {
    'SZ': '#3B82F6',          # blue
    'HC': '#F97316',          # orange
    'SZ_LIGHT': '#DBEAFE',
    'HC_LIGHT': '#FED7AA',
    'BG': '#F8FAFC',
    'BORDER': '#64748B',
    'ARROW': '#334155',
    'GREEN': '#22C55E',
    'RED': '#EF4444',
    'GRAY': '#94A3B8',
    'PURPLE': '#7C3AED',
    'DARK': '#1E293B',
    'AMBIG': '#9CA3AF',
    'SITE_A': '#10B981',
    'SITE_B': '#8B5CF6',
    'SITE_C': '#EC4899',
}

# — Color variables matching the standalone .py files —
# Use these so code copied from fig1_crf_process.py etc. works without edits.
C_SZ = COLORS['SZ']
C_HC = COLORS['HC']
C_BG = COLORS['BG']
C_BORDER = COLORS['BORDER']
C_ARROW = COLORS['ARROW']
C_GREEN = COLORS['GREEN']
C_RED = COLORS['RED']
C_GRAY = COLORS['GRAY']
C_PURPLE = COLORS['PURPLE']
C_LIGHT_BLUE = COLORS['SZ_LIGHT']
C_LIGHT_ORANGE = COLORS['HC_LIGHT']
C_SZ_LIGHT = COLORS['SZ_LIGHT']
# — Color variables matching the standalone .py files —
# Use these so code copied from fig1_crf_process.py etc. works without edits.
C_SZ = COLORS['SZ']
C_HC = COLORS['HC']
C_BG = COLORS['BG']
C_BORDER = COLORS['BORDER']
C_ARROW = COLORS['ARROW']
C_GREEN = COLORS['GREEN']
C_RED = COLORS['RED']
C_GRAY = COLORS['GRAY']
C_PURPLE = COLORS['PURPLE']
C_LIGHT_BLUE = COLORS['SZ_LIGHT']
C_LIGHT_ORANGE = COLORS['HC_LIGHT']
C_SZ_LIGHT = COLORS['SZ_LIGHT']
C_HC_LIGHT = COLORS['HC_LIGHT']
C_AMBIG = COLORS['AMBIG']
C_SITE_A = COLORS['SITE_A']
C_SITE_B = COLORS['SITE_B']
C_SITE_C = COLORS['SITE_C']


def create_canvas(width=16, height=10, xlim=None, ylim=None):
    """Create a blank canvas (fig, ax) with axis turned off.

    Args:
        width:  figure width in inches
        height: figure height in inches
        xlim:   tuple (xmin, xmax), defaults to (0, width)
        ylim:   tuple (ymin, ymax), defaults to (0, height)

    Returns:
        (fig, ax) ready for drawing
    """
    fig, ax = plt.subplots(figsize=(width, height))
    ax.set_xlim(xlim or (0, width))
    ax.set_ylim(ylim or (0, height))
    ax.axis('off')
    return fig, ax


def save_step_image(fig, filename, dpi=200):
    """Save the current figure as a PNG in the output directory.

    Args:
        fig:      matplotlib figure object
        filename: output file name (e.g. 'step1_input_data.png')
        dpi:      resolution (default 200)
    """
    path = os.path.join(OUTPUT_DIR, filename)
    fig.savefig(path, dpi=dpi, bbox_inches='tight', facecolor='white',
                edgecolor='none', pad_inches=0.2)
    plt.close(fig)
    print(f"Saved: {path}")


def create_subplot_canvas(rows=1, cols=1, width=16, height=10, **kwargs):
    """Create a figure with multiple subplots.

    Args:
        rows, cols: subplot grid dimensions
        width, height: figure size in inches
        **kwargs: passed to plt.subplots()

    Returns:
        (fig, axes) - axes is always a flat array even for single subplot
    """
    fig, axes = plt.subplots(rows, cols, figsize=(width, height), **kwargs)
    if rows * cols == 1:
        axes = np.array([axes])
    return fig, axes.flatten()


print(f"Output directory: {OUTPUT_DIR}")
print("Utilities loaded: create_canvas(), save_step_image(), create_subplot_canvas()")
print("Color vars loaded: C_SZ, C_HC, C_BORDER, C_ARROW, C_GREEN, C_RED, C_GRAY, C_PURPLE, etc.")
print("→ Code from fig*.py files can be pasted directly without renaming colors.")
```

```python
def create_input_data_figure():
    fig, ax = create_canvas(width=8, height=2, xlim=(0, 8), ylim=(0, 5))

    ax.text(4, 5, 'STEP 1: Input Dataset', ha='center', fontsize=14, fontweight='bold', color=COLORS['BORDER'])

    # Draw table
    table_x, table_y = 1.5, 3.2
    col_w = 1
    row_h = 0.35
    headers = ['Feat 1', 'Feat 2', '...', 'Feat F', 'Label']
    n_rows = 6
    labels = [1, 2, 1, 2, 1, 2]
    label_names = {1: "SZ", 2: "HC"}

    for j, h in enumerate(headers):
        rect = FancyBboxPatch((table_x + j*col_w, table_y + row_h), col_w, row_h,
                              boxstyle="round,pad=0.02", fc='#E2E8F0', ec=COLORS['BORDER'], lw=1)
        ax.add_patch(rect)
        ax.text(table_x + j*col_w + col_w/2, table_y + row_h + row_h/2, h,
                ha='center', va='center', fontsize=9, fontweight='bold')

    np.random.seed(42)
    for i in range(n_rows):
        for j in range(len(headers)):
            y_pos = table_y - i * row_h
            if j == len(headers) - 1:
                fc = COLORS['SZ_LIGHT'] if labels[i] == 1 else COLORS['HC_LIGHT']
                txt = labels[i]
            else:
                fc = 'white'
                txt = f"{np.random.rand():.2f}" if j != 2 else '...'
            rect = FancyBboxPatch((table_x + j*col_w, y_pos), col_w, row_h,
                                  boxstyle="round,pad=0.02", fc=fc, ec=COLORS['BORDER'], lw=0.5)
            ax.add_patch(rect)
            ax.text(table_x + j*col_w + col_w/2, y_pos + row_h/2, txt,
                    ha='center', va='center', fontsize=8, color=COLORS['DARK'])

    # Dimension annotations
    ax.annotate('', xy=(table_x, table_y + row_h + 0.6), xytext=(table_x + 4*col_w, table_y + row_h + 0.6),
                arrowprops=dict(arrowstyle='<->', color=COLORS['BORDER'], lw=1.5))
    ax.text(table_x + 2*col_w, table_y + row_h + 0.8, 'F features', ha='center', fontsize=10, color=COLORS['BORDER'])

    ax.annotate('', xy=(table_x - 0.3, table_y), xytext=(table_x - 0.3, table_y - (n_rows-1)*row_h),
                arrowprops=dict(arrowstyle='<->', color=COLORS['BORDER'], lw=1.5))
    ax.text(table_x - 0.7, table_y - (n_rows/2), 'N\nsubjects', ha='center', va='center',
            fontsize=10, color=COLORS['BORDER'])

    ax.text(4, 1, 'Columns = upper-triangle FNC features',
            ha='center', fontsize=9, color=COLORS['GRAY'], style='italic')
    print(table_y, n_rows, row_h)
    save_step_image(fig, 'step1_input_data_format.png')

create_input_data_figure()
```

```python
def build_crf_trees_image():
    fig, ax = create_canvas(width=16, height=6, xlim=(0, 16), ylim=(0, 6))

    ax.text(8, 5.7, 'STEP 3: Build 201 Complete Random Forest Trees (in parallel)',
            ha='center', fontsize=14, fontweight='bold', color=C_BORDER)

    # Draw 5 trees side by side
    tree_positions = [2.5, 5, 7.5, 10, 12.5]
    tree_labels = ['Tree 1', 'Tree 2', 'Tree 3', '...', 'Tree 201']
    base_y = 5.0

    for idx, (tx, tl) in enumerate(zip(tree_positions, tree_labels)):
        if idx == 3:
            ax.text(tx, base_y - 1.5, '...', ha='center', va='center', fontsize=20, color=C_GRAY)
            continue

        # Tree box
        tree_box = FancyBboxPatch((tx - 1.1, base_y - 2.8), 2.2, 2.5,
                                  boxstyle="round,pad=0.1", fc='white', ec=C_BORDER, lw=1)
        ax.add_patch(tree_box)
        ax.text(tx, base_y - 0.5, tl, ha='center', fontsize=9, fontweight='bold', color='#1E293B')

        # Root node
        ax.plot(tx, base_y - 1.0, 'o', color='#6366F1', markersize=9, zorder=5)

        # Branches
        ax.plot([tx, tx - 0.5], [base_y - 1.0, base_y - 1.6], '-', color=C_BORDER, lw=1.2)
        ax.plot([tx, tx + 0.5], [base_y - 1.0, base_y - 1.6], '-', color=C_BORDER, lw=1.2)

        # Level 2 nodes
        ax.plot(tx - 0.5, base_y - 1.6, 'o', color='#6366F1', markersize=7, zorder=5)
        ax.plot(tx + 0.5, base_y - 1.6, 'o', color='#6366F1', markersize=7, zorder=5)

        # Leaf branches
        ax.plot([tx - 0.5, tx - 0.7], [base_y - 1.6, base_y - 2.1], '-', color=C_BORDER, lw=1)
        ax.plot([tx - 0.5, tx - 0.3], [base_y - 1.6, base_y - 2.1], '-', color=C_BORDER, lw=1)
        ax.plot([tx + 0.5, tx + 0.3], [base_y - 1.6, base_y - 2.1], '-', color=C_BORDER, lw=1)
        ax.plot([tx + 0.5, tx + 0.7], [base_y - 1.6, base_y - 2.1], '-', color=C_BORDER, lw=1)

        # Leaf nodes
        for lx, lc in [(tx - 0.7, C_SZ), (tx - 0.3, C_HC), (tx + 0.3, C_SZ), (tx + 0.7, C_HC)]:
            ax.plot(lx, base_y - 2.1, 'o', color=lc, markersize=7, zorder=5)

        ax.text(tx, base_y - 2.5, 'Otsu split\nper feature', ha='center', fontsize=8, color=C_GRAY)

    # Explanation note
    note_box = FancyBboxPatch((1, 0.4), 14, 1.0,
                              boxstyle="round,pad=0.1", fc='#FEF2F2', ec='#6366F1', lw=1)
    ax.add_patch(note_box)
    ax.text(8, 1.1, 'Each tree randomly picks one feature per node → splits using Otsu threshold',
            ha='center', fontsize=10, color='#4338CA')
    ax.text(8, 0.7, 'Leaf label vs ancestor label checked → if label flips along path, subject marked as noise candidate',
            ha='center', fontsize=9, color='#4338CA')

    save_step_image(fig, 'step3_build_crf_trees.png')

build_crf_trees_image()
```

```python
def noise_decision_image():
    fig, ax = create_canvas(width=16, height=5, xlim=(0, 16), ylim=(0, 5))

    ax.text(8, 4.7, 'STEP 4: Majority Vote \u2192 Noise Decision',
            ha='center', fontsize=14, fontweight='bold', color=C_BORDER)

    # Decision matrix
    dm_x, dm_y = 3.5, 3.8
    dm_cw, dm_rh = 1.35, 0.4
    dm_headers = ['Subj', 'T1', 'T2', '...', 'T402', 'Noise?']
    dm_data = [
        ['S1', '0', '0', '...', '0', 'No'],
        ['S2', '1', '1', '...', '1', 'Yes'],
        ['S3', '0', '0', '...', '0', 'No'],
        ['S4', '1', '1', '...', '1', 'Yes'],
    ]

    for j, h in enumerate(dm_headers):
        rect = FancyBboxPatch((dm_x + j*dm_cw, dm_y), dm_cw, dm_rh,
                              boxstyle="round,pad=0.02", fc='#E2E8F0', ec=C_BORDER, lw=1)
        ax.add_patch(rect)
        ax.text(dm_x + j*dm_cw + dm_cw/2, dm_y + dm_rh/2, h,
                ha='center', va='center', fontsize=9, fontweight='bold')

    for i, row in enumerate(dm_data):
        for j, val in enumerate(row):
            y_pos = dm_y - (i + 1) * dm_rh
            if j == len(row) - 1:
                fc = '#FEE2E2' if val == 'Yes' else '#DCFCE7'
                fw = 'bold'
            elif j > 0 and j < len(row) - 1 and val == '1':
                fc = '#FEE2E2'
                fw = 'bold'
            else:
                fc = 'white'
                fw = 'normal'
            rect = FancyBboxPatch((dm_x + j*dm_cw, y_pos), dm_cw, dm_rh,
                                  boxstyle="round,pad=0.02", fc=fc, ec=C_BORDER, lw=0.5)
            ax.add_patch(rect)
            ax.text(dm_x + j*dm_cw + dm_cw/2, y_pos + dm_rh/2, val,
                    ha='center', va='center', fontsize=9, fontweight=fw)

    ax.text(dm_x + 3*dm_cw, dm_y - 5.3*dm_rh,
            '0 = non-noise, 1 = noise\nNoise if > 50% of 402 trees say "noise"',
            ha='center', fontsize=9, color=C_GRAY, style='italic')

    save_step_image(fig, 'step4_noise_decision.png')

noise_decision_image()
```

```python
def identify_typical_subjects_image():
    np.random.seed(42)
    fig, ax = plt.subplots(figsize=(10, 5))

    n_subjects = 20
    ratios = np.clip(np.random.beta(5, 2, n_subjects), 0, 1)
    ratios = np.round(ratios, 1)
    ratios[3] = 0.4; ratios[7] = 0.5; ratios[11] = 0.3; ratios[15] = 0.6; ratios[18] = 0.2

    labels_orig = np.array([1, 2, 1, 2, 1, 2, 1, 2, 1, 2, 1, 2, 1, 2, 1, 2, 1, 2, 1, 2])
    threshold = 0.8

    colors = []
    for i, r in enumerate(ratios):
        if r >= threshold:
            colors.append(C_SZ if labels_orig[i] == 1 else C_HC)
        else:
            colors.append('#D1D5DB')

    ax.bar(range(n_subjects), ratios, color=colors, edgecolor='white', lw=0.5)
    ax.axhline(y=threshold, color=C_PURPLE, lw=2, ls='--', zorder=10)
    ax.text(n_subjects - 0.5, threshold + 0.03, f'Typical Threshold = {threshold}',
            ha='right', fontsize=10, color=C_PURPLE, fontweight='bold')

    ax.set_xlabel('Subject ID', fontsize=10)
    ax.set_ylabel('Non-noise Ratio (from CRF)', fontsize=10)
    ax.set_title('Identify Typical Subjects', fontsize=14, fontweight='bold', color=C_BORDER)
    ax.set_ylim(0, 1.15)
    ax.set_xticks(range(0, n_subjects, 2))
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    typical_count = sum(1 for r in ratios if r >= threshold)
    atypical_count = n_subjects - typical_count
    ax.text(0.02, 0.95, f'Typical: {typical_count} subjects (colored)',
            transform=ax.transAxes, fontsize=9, color='#1E293B',
            bbox=dict(boxstyle='round,pad=0.3', fc='#F0FDF4', ec=C_GREEN))
    ax.text(0.02, 0.85, f'Atypical: {atypical_count} subjects (gray)',
            transform=ax.transAxes, fontsize=9, color='#1E293B',
            bbox=dict(boxstyle='round,pad=0.3', fc='#F3F4F6', ec='#9CA3AF'))

    save_step_image(fig, 'fig2a_identify_typical.png')

identify_typical_subjects_image()
```

```python
def feature_selection_image():
    np.random.seed(42)
    fig, ax = plt.subplots(figsize=(10, 5))

    n_features = 15
    p_values = np.random.exponential(0.01, n_features)
    p_values[2] = 0.5; p_values[5] = 0.3; p_values[8] = 0.15; p_values[12] = 0.4
    bonferroni_threshold = 0.01 / n_features

    bar_colors = [C_GREEN if p < bonferroni_threshold else '#D1D5DB' for p in p_values]
    ax.bar(range(n_features), -np.log10(p_values), color=bar_colors, edgecolor='white', lw=0.5)
    ax.axhline(y=-np.log10(bonferroni_threshold), color=C_PURPLE, lw=2, ls='--')
    ax.text(n_features - 0.5, -np.log10(bonferroni_threshold) + 0.1,
            f'Bonferroni: 0.01/{n_features}',
            ha='right', fontsize=9, color=C_PURPLE, fontweight='bold')

    selected = sum(1 for p in p_values if p < bonferroni_threshold)
    ax.set_xlabel('Feature Index', fontsize=10)
    ax.set_ylabel('-log10(p-value)', fontsize=10)
    ax.set_title('Bonferroni-corrected t-test Feature Selection', fontsize=14, fontweight='bold', color=C_BORDER)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.text(0.02, 0.95, f'{selected} features selected (green bars)',
            transform=ax.transAxes, fontsize=9, color='#1E293B',
            bbox=dict(boxstyle='round,pad=0.3', fc='#F0FDF4', ec=C_GREEN))

    save_step_image(fig, 'fig2b_feature_selection.png')

feature_selection_image()
```

```python
def compute_centroids_image():
    np.random.seed(42)
    fig, ax = plt.subplots(figsize=(10, 7))
    n_typ_sz, n_typ_hc = 12, 10

    sz_x = np.random.normal(-1.5, 0.5, n_typ_sz)
    sz_y = np.random.normal(1.0, 0.5, n_typ_sz)
    hc_x = np.random.normal(1.5, 0.6, n_typ_hc)
    hc_y = np.random.normal(-0.5, 0.5, n_typ_hc)

    ax.scatter(sz_x, sz_y, c=C_SZ, s=70, alpha=0.5, edgecolors='white', lw=0.5, zorder=3)
    ax.scatter(hc_x, hc_y, c=C_HC, s=70, alpha=0.5, edgecolors='white', lw=0.5, zorder=3)

    # Ellipses
    ell_sz = Ellipse(xy=(np.mean(sz_x), np.mean(sz_y)),
                     width=3*np.std(sz_x), height=3*np.std(sz_y),
                     angle=0, fc='none', ec=C_SZ, lw=1.5, ls='--', alpha=0.5)
    ell_hc = Ellipse(xy=(np.mean(hc_x), np.mean(hc_y)),
                     width=3*np.std(hc_x), height=3*np.std(hc_y),
                     angle=0, fc='none', ec=C_HC, lw=1.5, ls='--', alpha=0.5)
    ax.add_patch(ell_sz)
    ax.add_patch(ell_hc)

    # Centroids
    c_sz_x, c_sz_y = np.mean(sz_x), np.mean(sz_y)
    c_hc_x, c_hc_y = np.mean(hc_x), np.mean(hc_y)

    ax.scatter([c_sz_x], [c_sz_y], c=C_SZ, s=400, marker='*', edgecolors='#1E40AF',
               lw=2, zorder=10, label='Centroid SZ')
    ax.scatter([c_hc_x], [c_hc_y], c=C_HC, s=400, marker='*', edgecolors='#C2410C',
               lw=2, zorder=10, label='Centroid HC')

    # Lines from points to centroids
    for x, y in zip(sz_x[:4], sz_y[:4]):
        ax.plot([x, c_sz_x], [y, c_sz_y], '--', color=C_SZ, alpha=0.2, lw=0.8)
    for x, y in zip(hc_x[:4], hc_y[:4]):
        ax.plot([x, c_hc_x], [y, c_hc_y], '--', color=C_HC, alpha=0.2, lw=0.8)

    ax.annotate('Centroid SZ\n(mean of typical SZ)', xy=(c_sz_x, c_sz_y),
                xytext=(c_sz_x + 1.2, c_sz_y + 0.8),
                fontsize=9, color=C_SZ, fontweight='bold',
                arrowprops=dict(arrowstyle='->', color=C_SZ, lw=1.5))
    ax.annotate('Centroid HC\n(mean of typical HC)', xy=(c_hc_x, c_hc_y),
                xytext=(c_hc_x - 1.2, c_hc_y - 1.0),
                fontsize=9, color=C_HC, fontweight='bold',
                arrowprops=dict(arrowstyle='->', color=C_HC, lw=1.5))

    ax.set_xlabel('Selected Feature 1', fontsize=10)
    ax.set_ylabel('Selected Feature 2', fontsize=10)
    ax.set_title('Compute Centroids (means) in Selected Feature Space',
                 fontsize=14, fontweight='bold', color=C_BORDER)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    # output_text = 'Output per set to server:  - Centroid SZ (mean vector)  - Centroid HC (mean vector)  - Selected feature indices'
    # ax.text(0.02, 0.02, output_text, transform=ax.transAxes, fontsize=9,
    #         verticalalignment='bottom', color='#1E293B',
    #         bbox=dict(boxstyle='round,pad=0.4', fc='#EEF2FF', ec='#6366F1', lw=1))

    save_step_image(fig, 'fig2c_compute_centroids.png')

compute_centroids_image()
```

```python
def score_formula_image():
    fig, ax = create_canvas(width=10, height=8, xlim=(0, 10), ylim=(0, 10))
    ax.text(5, 9.7, 'Compute Dimensional Score', ha='center', fontsize=14, fontweight='bold', color=C_BORDER)

    # Formula box
    formula_box = FancyBboxPatch((0.5, 7.2), 9, 2.2, boxstyle="round,pad=0.2", fc='#EEF2FF', ec='#6366F1', lw=1.5)
    ax.add_patch(formula_box)
    formulas = [('A = d$_{SZ}$ / (d$_{SZ}$ + d$_{HC}$)', 8.9),
                ('B = d$_{HC}$ / (d$_{SZ}$ + d$_{HC}$)', 8.4),
                ('score = tanh ((A - B) * \\pi/5 / 2 )', 7.9)]
    for txt, y in formulas:
        ax.text(5, y, txt, ha='center', fontsize=14, color='#4338CA', fontweight='bold', family='serif')

    ax.text(5, 7.0, 'Interpretation', ha='center', fontsize=11, fontweight='bold', color='#1E293B')

    # Number line
    nl_y = 5.8
    nl_x_start, nl_x_end = 1, 9
    ax.annotate('', xy=(nl_x_end, nl_y), xytext=(nl_x_start, nl_y),
                arrowprops=dict(arrowstyle='->', color='#1E293B', lw=2))
    ax.annotate('', xy=(nl_x_start, nl_y), xytext=(nl_x_end, nl_y),
                arrowprops=dict(arrowstyle='->', color='#1E293B', lw=2))
    zero_x = 5
    ax.plot([zero_x, zero_x], [nl_y - 0.2, nl_y + 0.2], '-', color='#1E293B', lw=2)
    ax.text(zero_x, nl_y - 0.4, '0', ha='center', fontsize=10, color='#1E293B')
    ax.fill_between([nl_x_start, zero_x], nl_y - 0.15, nl_y + 0.15, color=C_SZ, alpha=0.15)
    ax.text(nl_x_start + 0.3, nl_y - 0.6, '-inf', ha='left', fontsize=9, color=C_SZ, fontweight='bold')
    ax.text((nl_x_start + zero_x) / 2, nl_y - 0.4, 'Closer to SZ centroid', ha='center', fontsize=9, color=C_SZ)
    ax.fill_between([zero_x, nl_x_end], nl_y - 0.15, nl_y + 0.15, color=C_HC, alpha=0.15)
    ax.text(nl_x_end - 0.3, nl_y - 0.6, '+inf', ha='right', fontsize=9, color=C_HC, fontweight='bold')
    ax.text((zero_x + nl_x_end) / 2, nl_y - 0.4, 'Closer to HC centroid', ha='center', fontsize=9, color=C_HC)

    example_score = 3.5
    ax.plot(example_score, nl_y, 'o', color='#1E293B', markersize=10, zorder=10)
    ax.text(example_score, nl_y - 0.8, 'Subject score', ha='center', fontsize=9, color='#1E293B', fontweight='bold')

    # Averaging table
    ax.text(5, 4.7, 'Average Scores from All External Sites', ha='center', fontsize=13, fontweight='bold', color=C_BORDER)
    table_data = [['Subject', 'Score\n(from B)', 'Score\n(from C)', 'Average'],
                  ['S1', '-0.85', '-0.72', '-0.79'],
                  ['S2', '+1.23', '+1.05', '+1.14'],
                  ['S3', '-0.12', '+0.08', '-0.02']]
    t_x, t_y = 1.5, 2.6
    t_cw, t_rh = 2.0, 0.65

    for j, h in enumerate(table_data[0]):
        rect = FancyBboxPatch((t_x + j*t_cw, t_y + t_rh), t_cw, t_rh, boxstyle="round,pad=0.02", fc='#E2E8F0', ec=C_BORDER, lw=1)
        ax.add_patch(rect)
        ax.text(t_x + j*t_cw + t_cw/2, t_y + t_rh + t_rh/2, h, ha='center', va='center', fontsize=9, fontweight='bold')

    for i, row in enumerate(table_data[1:]):
        for j, val in enumerate(row):
            y_pos = t_y - i*t_rh
            if j == 3:
                fc = '#EEF2FF'; fw = 'bold'
            elif j == 1:
                fc = '#F5F3FF'; fw = 'normal'
            elif j == 2:
                fc = '#FDF2F8'; fw = 'normal'
            else:
                fc = 'white'; fw = 'normal'
            rect = FancyBboxPatch((t_x + j*t_cw, y_pos), t_cw, t_rh, boxstyle="round,pad=0.02", fc=fc, ec=C_BORDER, lw=0.5)
            ax.add_patch(rect)
            ax.text(t_x + j*t_cw + t_cw/2, y_pos + t_rh/2, val, ha='center', va='center', fontsize=10, fontweight=fw)

    ax.text(t_x + 1*t_cw + t_cw/2, t_y - 2*t_rh - 0.35, 'Site B', ha='center', fontsize=9, color=C_SITE_B, fontweight='bold')
    ax.text(t_x + 2*t_cw + t_cw/2, t_y - 2*t_rh - 0.35, 'Site C', ha='center', fontsize=9, color=C_SITE_C, fontweight='bold')
    ax.text(5, 0.8, 'Note: A site skips its own centroids — only scores against other sites.', ha='center', fontsize=9, color=C_GRAY, style='italic')
    ax.text(5, 0.3, 'The average score is sent to the server for adaptive thresholding.', ha='center', fontsize=9, color='#1E293B',
            bbox=dict(boxstyle='round,pad=0.3', fc='#F0FDF4', ec=C_GREEN, lw=1))

    save_step_image(fig, 'fig3c_score_formula.png')

score_formula_image()
```

```python
def sorted_truncation_image():
    np.random.seed(42)
    fig, ax = plt.subplots(figsize=(14, 6))
    scores_neg = np.sort(-np.abs(np.random.randn(30) * 1.0 + 1.2))
    scores_pos = np.sort(np.abs(np.random.randn(30) * 1.0 + 1.2))
    all_scores = np.sort(np.concatenate([scores_neg, scores_pos]))
    n_total = len(all_scores)
    colors = [C_SZ if s < 0 else C_HC for s in all_scores]
    ax.bar(range(n_total), all_scores, color=colors, edgecolor='white', lw=0.3, width=0.8)

    N_neg = sum(1 for s in all_scores if s < 0)
    N_pos = sum(1 for s in all_scores if s > 0)
    trunc = 0.2
    i_neg = int(round(N_neg * (1 - trunc))) - 1
    i_pos = N_neg + int(round(N_pos * trunc)) - 1

    ax.axvline(x=i_neg, color=C_SZ, lw=2.5, ls='--', zorder=10)
    ax.axvline(x=i_pos, color=C_HC, lw=2.5, ls='--', zorder=10)

    s_neg_val = all_scores[i_neg]
    s_pos_val = all_scores[i_pos]

    ax.annotate(f's_neg = {s_neg_val:.2f}\n(80th %ile of negatives)',
                xy=(i_neg, s_neg_val), xytext=(i_neg - 15, s_neg_val - 1.5),
                fontsize=9, color=C_SZ, fontweight='bold',
                arrowprops=dict(arrowstyle='->', color=C_SZ, lw=1.5))

    ax.annotate(f's_pos = {s_pos_val:.2f}\n(20th %ile of positives)',
                xy=(i_pos, s_pos_val), xytext=(i_pos + 5, s_pos_val + 1.2),
                fontsize=9, color=C_HC, fontweight='bold',
                arrowprops=dict(arrowstyle='->', color=C_HC, lw=1.5))

    ax.axhline(y=0, color='#1E293B', lw=1, ls='--', alpha=0.3)
    ax.set_xlabel('Subjects (sorted by score)', fontsize=11)
    ax.set_ylabel('Dimensional Score', fontsize=11)
    ax.set_title('Sort All Scores and Apply Truncation (TruncationParameter = 0.2)',
                 fontsize=14, fontweight='bold', color=C_BORDER)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    ax.text(0.02, 0.95, f'N_neg = {N_neg}', transform=ax.transAxes,
            fontsize=10, color=C_SZ, fontweight='bold')
    ax.text(0.15, 0.95, f'N_pos = {N_pos}', transform=ax.transAxes,
            fontsize=10, color=C_HC, fontweight='bold')

    t_val = abs(s_neg_val + s_pos_val) / 2
    t_display = t_val * np.pi / 2.5
    ax.text(0.98, 0.95, f'(function) transform: Transform / 2 = {t_display:.2f}',
            transform=ax.transAxes, ha='right',
            fontsize=11, color=C_PURPLE, fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.3', fc='#F5F3FF', ec=C_PURPLE, lw=1.5))

    save_step_image(fig, 'fig4b_sorted_truncation.png')

sorted_truncation_image()
```

```python
def relabeling_image():
    fig, ax = plt.subplots(figsize=(14, 5))
    t_val = 0.65  # illustrative threshold
    ax.set_xlim(-4, 4)
    ax.set_ylim(-0.8, 2.8)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.set_yticks([])
    ax.set_title('Apply Threshold to Relabel Subjects', fontsize=14, fontweight='bold', color=C_BORDER, pad=15)

    # Colored zones
    ax.axvspan(-4, -t_val, alpha=0.15, color=C_SZ, zorder=0)
    ax.axvspan(-t_val, t_val, alpha=0.15, color=C_AMBIG, zorder=0)
    ax.axvspan(t_val, 4, alpha=0.15, color=C_HC, zorder=0)

    # Threshold lines
    ax.axvline(x=-t_val, color=C_RED, lw=3, ls='--', zorder=10)
    ax.axvline(x=t_val, color=C_RED, lw=3, ls='--', zorder=10)
    ax.axhline(y=0, color='#1E293B', lw=1, ls='--', alpha=0.3)

    # Zone labels
    ax.text(-t_val - (4 - t_val)/2, 2.4, 'Relabeled SZ\n(score < -t)', ha='center', fontsize=12, fontweight='bold', color=C_SZ,
            bbox=dict(boxstyle='round,pad=0.3', fc=C_SZ, ec='none', alpha=0.15))
    ax.text(0, 2.4, 'Ambiguous\n(label = -1)', ha='center', fontsize=12, fontweight='bold', color=C_AMBIG,
            bbox=dict(boxstyle='round,pad=0.3', fc=C_AMBIG, ec='none', alpha=0.15))
    ax.text(t_val + (4 - t_val)/2, 2.4, 'Relabeled HC\n(score > +t)', ha='center', fontsize=12, fontweight='bold', color=C_HC,
            bbox=dict(boxstyle='round,pad=0.3', fc=C_HC, ec='none', alpha=0.15))

    # Threshold labels
    ax.text(-t_val, -0.55, f'-t = {-t_val:.2f}', ha='center', fontsize=11, color=C_RED, fontweight='bold')
    ax.text(t_val, -0.55, f'+t = {t_val:.2f}', ha='center', fontsize=11, color=C_RED, fontweight='bold')
    ax.text(0, -0.55, '0', ha='center', fontsize=10, color='#1E293B')

    # Scatter subjects
    np.random.seed(123)
    subject_scores = np.random.randn(30) * 1.2
    subject_y = np.random.uniform(0.1, 1.5, 30)
    for s, y in zip(subject_scores, subject_y):
        if s < -t_val:
            c = C_SZ
            marker = 'o'
        elif s > t_val:
            c = C_HC
            marker = 'o'
        else:
            c = C_AMBIG
            marker = 'x'
        ax.plot(s, y, marker, color=c, markersize=8, alpha=0.7, markeredgewidth=1.5)

    # Legend
    legend_elements = [
        mpatches.Patch(facecolor=C_SZ, alpha=0.3, label='Relabeled as SZ (label=2)'),
        mpatches.Patch(facecolor=C_HC, alpha=0.3, label='Relabeled as HC (label=1)'),
        mpatches.Patch(facecolor=C_AMBIG, alpha=0.3, label='Ambiguous (label=-1)'),
        plt.Line2D([0], [0], color=C_RED, lw=3, ls='--', label=f'Threshold (± {t_val:.2f})')
    ]
    ax.legend(handles=legend_elements, loc='lower right', fontsize=9, ncol=2, frameon=True, fancybox=True)
    ax.set_xlabel('Dimensional Score', fontsize=12)

    save_step_image(fig, 'fig4c_relabeling.png')

relabeling_image()
```
