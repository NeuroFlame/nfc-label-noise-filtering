```python
import numpy as np
import os

def save_fnc_matrices(fnc_array, out_dir="fnc"):
    """
    fnc_array: numpy array of shape (N, P, P)
    out_dir: folder to save the FNC matrices
    """
    os.makedirs(out_dir, exist_ok=True)

    N = fnc_array.shape[0]

    for i in range(N):
        matrix = fnc_array[i]                       # shape (P, P)
        filename = os.path.join(out_dir, f"sub_{i+1:03d}.npy")
        np.save(filename, matrix)

    print(f"Saved {N} FNC matrices to '{out_dir}/'")

import csv

def create_labels_csv(labels, csv_path="subjects.csv"):
    """
    labels: array-like of length N containing labels (e.g., [1,2,1,1,...])
    csv_path: where to save the CSV file
    """

    with open(csv_path, mode="w", newline="") as f:
        writer = csv.writer(f)

        # Write header
        writer.writerow(["subject_file", "label"])

        # Write each row
        for i, label in enumerate(labels):
            filename = f"sub_{i+1:03d}.npy"     # sub_001.npy, sub_002.npy, ...
            writer.writerow([filename, label])

    print(f"CSV saved at: {csv_path}")


for site in ['site1', 'site2']:
    mat = os.path.join(site, 'data.mat')
    dataset = load_data_matfile(mat, name=[
            'sFNC',
            'FILE_ID',
            'analysis_SCORE',
        ])
    file_ids = dataset['FILE_ID']
    label_index = next(
        (i for i, col in enumerate(file_ids) if "diagnosis" in col.lower()),
        None
    )
    if label_index is None:
        raise ValueError(f'No "diagnosis" column found in FILE_ID for')

    labels = dataset['analysis_SCORE'][:, label_index]
    labels = [int(i) for i in labels]
    create_labels_csv(labels, os.path.join(site, 'subjects.csv'))
```
